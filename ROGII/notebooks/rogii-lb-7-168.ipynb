```python
SELECTOR_N_EVAL_THRESHOLD = 4840.0
SELECTOR_Z_SPAN_THRESHOLDS = (136.73000000000016, 185.5133333333342)
SELECTOR_BIN_VARIANTS = {0: 'pf_scale_5_hold_0.2', 1: 'pf_scale_3_hold_0.15', 2: 'pf_scale_12_beam_0.2_hold_0.15', 3: 'pf_scale_5_hold_0.15', 4: 'pf_scale_5_beam_0.05_hold_0.05', 5: 'pf_scale_12_beam_0.2_hold_0.05'}
SELECTOR_GLOBAL_VARIANT = 'pf_scale_8_hold_0.2'
SELECTOR_SCALES = (3.0, 5.0, 8.0, 12.0)
R2_FORMATION_COLS_004 = ['ANCC', 'ASTNU', 'ASTNL', 'EGFDU', 'EGFDL', 'BUDA']
R2_BEAM_CONFIGS_005 = [(10, 20.0, 144.0, 2), (10, 8.0, 64.0, 2), (8, 35.0, 220.0, 1), (10, 14.0, 90.0, 5), (20, 4.0, 36.0, 3), (12, 12.0, 100.0, 3), (15, 25.0, 180.0, 2), (20, 30.0, 200.0, 2), (15, 10.0, 80.0, 4), (25, 6.0, 50.0, 3), (10, 40.0, 300.0, 1), (12, 18.0, 120.0, 5), (30, 8.0, 70.0, 2), (10, 50.0, 400.0, 0)]

def r2_tvt_from_contacts_006(hw_tr, tw_tr, ref_col='EGFDU'):
    r2_tw_g_007 = tw_tr.dropna(subset=['Geology'])
    r2_ref_tvt_008 = r2_tw_g_007[r2_tw_g_007['Geology'] == ref_col]['TVT'].min()
    if np.isnan(r2_ref_tvt_008):
        ref_col = r2_tw_g_007['Geology'].iloc[0]
        r2_ref_tvt_008 = r2_tw_g_007[r2_tw_g_007['Geology'] == ref_col]['TVT'].min()
    r2_offset_009 = (hw_tr['TVT'] - (r2_ref_tvt_008 - (hw_tr['Z'] - hw_tr[ref_col]))).mean()
    return r2_ref_tvt_008 - (hw_tr['Z'] - hw_tr[ref_col]) + r2_offset_009

def r2_load_well_010(wid, split='train'):
    r2_base_011 = CFG.dataset_path / split
    hw = pd.read_csv(r2_base_011 / f'{wid}__horizontal_well.csv')
    tw = pd.read_csv(r2_base_011 / f'{wid}__typewell.csv')
    return (hw, tw)

def r2_run_particle_filter_012(hw, tw, n_particles=500, seed=42):
    r2_tw_s_013 = tw.sort_values('TVT')
    tw_tvt = r2_tw_s_013['TVT'].values.astype(float)
    tw_gr = r2_tw_s_013['GR'].fillna(r2_tw_s_013['GR'].mean()).values.astype(float)
    r2_kn_slot_014 = hw[hw['TVT_input'].notna()]
    r2_ev_slot_015 = hw[hw['TVT_input'].isna()]
    if len(r2_ev_slot_015) == 0:
        return (hw['TVT_input'].values.astype(float).copy(), 0.0)
    r2_last_016 = r2_kn_slot_014.iloc[-1]
    last_tvt = float(r2_last_016['TVT_input'])
    r2_last_z_017 = float(r2_last_016['Z'])
    r2_last_md_018 = float(r2_last_016['MD'])
    tw_at_k = np.interp(r2_kn_slot_014['TVT_input'].values, tw_tvt, tw_gr)
    gs = float(np.clip(np.nanstd(r2_kn_slot_014['GR'].fillna(0).values - tw_at_k), 10.0, 60.0))
    r2_tail_019 = r2_kn_slot_014.tail(30)
    r2_dt_slot_020 = np.diff(r2_tail_019['TVT_input'].values)
    r2_dz_slot_021 = np.diff(r2_tail_019['Z'].values)
    r2_dm_slot_022 = np.diff(r2_tail_019['MD'].values)
    r2_m_slot_023 = r2_dm_slot_022 > 0
    ir = float(np.median((r2_dt_slot_020 + r2_dz_slot_021)[r2_m_slot_023] / r2_dm_slot_022[r2_m_slot_023])) if r2_m_slot_023.sum() >= 3 else 0.0
    N = n_particles
    r2_rng_024 = np.random.default_rng(seed)
    ls = last_tvt + r2_last_z_017
    pos = ls + 4.5 * r2_rng_024.standard_normal(N)
    r2_rate_025 = ir + 0.01 * r2_rng_024.standard_normal(N)
    w = np.ones(N) / N
    MOM = 0.998
    VN = 0.002
    PN = 0.005
    RP = 0.1
    RR = 0.001
    RESAMP = 0.5
    md_v = r2_ev_slot_015['MD'].values.astype(float)
    z_v = r2_ev_slot_015['Z'].values.astype(float)
    r2_gr_interp_026 = hw['GR'].interpolate(limit_direction='both').fillna(tw_gr.mean())
    gr_v = r2_gr_interp_026.values.astype(float)[r2_ev_slot_015.index]
    r2_out_vals_027 = hw['TVT_input'].values.astype(float).copy()
    r2_res_028 = np.empty(len(r2_ev_slot_015))
    r2_prev_md_029 = r2_last_md_018
    r2_log_lik_030 = 0.0
    for r2_i_slot_031 in range(len(r2_ev_slot_015)):
        r2_dm_step_032 = max(md_v[r2_i_slot_031] - r2_prev_md_029, 1.0)
        r2_rate_025 = MOM * r2_rate_025 + VN * r2_rng_024.standard_normal(N)
        pos = pos + r2_rate_025 * r2_dm_step_032 + PN * r2_rng_024.standard_normal(N)
        r2_tvt_p_033 = pos - z_v[r2_i_slot_031]
        r2_tvt_p_033 = np.clip(r2_tvt_p_033, tw_tvt[0] - 100, tw_tvt[-1] + 100)
        pos = r2_tvt_p_033 + z_v[r2_i_slot_031]
        r2_eg_slot_034 = np.interp(r2_tvt_p_033, tw_tvt, tw_gr)
        r2_d_slot_035 = (gr_v[r2_i_slot_031] - r2_eg_slot_034) / gs
        r2_lk_slot_036 = np.exp(-0.5 * np.minimum(r2_d_slot_035 ** 2, 600.0))
        r2_lk_slot_036 = np.maximum(r2_lk_slot_036, 1e-300)
        r2_avg_lk_037 = float((w * r2_lk_slot_036).sum())
        r2_log_lik_030 += np.log(max(r2_avg_lk_037, 1e-300))
        w = w * r2_lk_slot_036
        r2_ws_slot_038 = w.sum()
        w = w / r2_ws_slot_038 if r2_ws_slot_038 > 0 else np.ones(N) / N
        r2_n_eff_039 = 1.0 / (w ** 2).sum()
        if r2_n_eff_039 < RESAMP * N:
            r2_cum_040 = np.cumsum(w)
            r2_u0_slot_041 = r2_rng_024.uniform(0, 1.0 / N)
            r2_idx_042 = np.clip(np.searchsorted(r2_cum_040, r2_u0_slot_041 + np.arange(N) / N), 0, N - 1)
            pos = pos[r2_idx_042] + RP * r2_rng_024.standard_normal(N)
            r2_rate_025 = r2_rate_025[r2_idx_042] + RR * r2_rng_024.standard_normal(N)
            w = np.ones(N) / N
        r2_res_028[r2_i_slot_031] = float(np.dot(w, pos - z_v[r2_i_slot_031]))
        r2_prev_md_029 = md_v[r2_i_slot_031]
    r2_out_vals_027[list(r2_ev_slot_015.index)] = r2_res_028
    return (r2_out_vals_027, r2_log_lik_030)

def r2_run_pf_lik_ensemble_043(hw, tw, n_particles=500, n_seeds=128, scale=5.0):
    r2_preds_044 = []
    r2_liks_045 = []
    for s in range(n_seeds):
        p, r2_ll_slot_046 = r2_run_particle_filter_012(hw, tw, n_particles=n_particles, seed=s)
        r2_preds_044.append(p)
        r2_liks_045.append(r2_ll_slot_046)
    r2_liks_045 = np.array(r2_liks_045)
    r2_liks_n_047 = r2_liks_045 - r2_liks_045.max()
    r2_weights_048 = np.exp(r2_liks_n_047 / scale)
    r2_weights_048 /= r2_weights_048.sum()
    return (r2_weights_048[:, None] * np.stack(r2_preds_044, 0)).sum(0)

def run_pf_lik_ensemble_scales(hw, tw, scales=SELECTOR_SCALES, n_particles=500, n_seeds=128):
    r2_preds_044 = []
    r2_liks_045 = []
    for s in range(n_seeds):
        p, r2_ll_slot_046 = r2_run_particle_filter_012(hw, tw, n_particles=n_particles, seed=s)
        r2_preds_044.append(p)
        r2_liks_045.append(r2_ll_slot_046)
    r2_pred_arr_049 = np.stack(r2_preds_044, 0)
    r2_liks_045 = np.array(r2_liks_045)
    r2_liks_n_047 = r2_liks_045 - r2_liks_045.max()
    r2_out_050 = {}
    for scale in scales:
        r2_weights_048 = np.exp(r2_liks_n_047 / float(scale))
        r2_weights_048 /= r2_weights_048.sum()
        r2_out_050[f'pf_scale_{scale:g}'] = (r2_weights_048[:, None] * r2_pred_arr_049).sum(0)
    r2_out_050['pf_mean'] = r2_pred_arr_049.mean(0)
    return r2_out_050

def r2_beam_search_051(hgr, tw_tvt, tw_gr, last_tvt, bs=10, mc=20.0, es=144.0, r=2):
    r2_n_slot_052 = len(hgr)
    r2_nt_slot_053 = len(tw_tvt)
    if r2_n_slot_052 == 0:
        return np.array([last_tvt])
    if r > 0 and r2_n_slot_052 > max(3, 2 * r + 1):
        r2_win_054 = min(2 * r + 1, r2_n_slot_052 if r2_n_slot_052 % 2 == 1 else r2_n_slot_052 - 1)
        sgr = savgol_filter(hgr, r2_win_054, min(2, r2_win_054 - 1))
    else:
        sgr = hgr.copy()
    si = int(np.argmin(np.abs(tw_tvt - last_tvt)))
    R2_MOVES_055 = np.array([-2, -1, 0, 1, 2], dtype=np.int64)
    R2_MC_056 = mc * np.array([2.0, 1.0, 0.0, 1.0, 2.0])
    r2_bidx_057 = np.full(bs, si, dtype=np.int64)
    r2_bcost_058 = np.full(bs, np.inf)
    r2_bcost_058[0] = 0.0
    r2_bn_slot_059 = 1
    r2_result_060 = np.zeros(r2_n_slot_052)
    for step in range(r2_n_slot_052):
        r2_gv_slot_061 = sgr[step]
        r2_ni_slot_062 = r2_bidx_057[:r2_bn_slot_059, None] + R2_MOVES_055[None, :]
        r2_ci_slot_063 = np.clip(r2_ni_slot_062, 0, r2_nt_slot_053 - 1)
        r2_valid_064 = (r2_ni_slot_062 >= 0) & (r2_ni_slot_062 < r2_nt_slot_053)
        r2_gr_e_065 = (r2_gv_slot_061 - tw_gr[r2_ci_slot_063]) ** 2 / es
        r2_tot_066 = r2_bcost_058[:r2_bn_slot_059, None] + r2_gr_e_065 + R2_MC_056[None, :]
        r2_tot_066 = np.where(r2_valid_064, r2_tot_066, np.inf)
        r2_ni_f_067 = r2_ni_slot_062.flatten()
        r2_tot_f_068 = r2_tot_066.flatten()
        r2_vf_slot_069 = r2_valid_064.flatten()
        r2_ni_f_067 = r2_ni_f_067[r2_vf_slot_069]
        r2_tot_f_068 = r2_tot_f_068[r2_vf_slot_069]
        r2_order_070 = np.argsort(r2_tot_f_068)
        r2_ni_s_071 = r2_ni_f_067[r2_order_070]
        r2_tot_s_072 = r2_tot_f_068[r2_order_070]
        _, r2_first_073 = np.unique(r2_ni_s_071, return_index=True)
        r2_ni_u_074 = r2_ni_s_071[r2_first_073]
        r2_tot_u_075 = r2_tot_s_072[r2_first_073]
        r2_kept_076 = min(bs, len(r2_ni_u_074))
        r2_top_077 = np.argpartition(r2_tot_u_075, min(r2_kept_076 - 1, len(r2_tot_u_075) - 1))[:r2_kept_076]
        r2_top_077 = r2_top_077[np.argsort(r2_tot_u_075[r2_top_077])]
        r2_bidx_057[:r2_kept_076] = r2_ni_u_074[r2_top_077]
        r2_bcost_058[:r2_kept_076] = r2_tot_u_075[r2_top_077]
        if r2_kept_076 < bs:
            r2_bidx_057[r2_kept_076:] = r2_bidx_057[r2_kept_076 - 1]
            r2_bcost_058[r2_kept_076:] = np.inf
        r2_bn_slot_059 = r2_kept_076
        r2_result_060[step] = tw_tvt[r2_bidx_057[0]]
    return r2_result_060

def run_beam_ensemble(hw, tw):
    r2_kn_slot_014 = hw[hw['TVT_input'].notna()]
    r2_ev_slot_015 = hw[hw['TVT_input'].isna()]
    if len(r2_ev_slot_015) == 0:
        return hw['TVT_input'].values.astype(float).copy()
    last_tvt = float(r2_kn_slot_014.iloc[-1]['TVT_input'])
    r2_tw_s_013 = tw.sort_values('TVT')
    tw_tvt = r2_tw_s_013['TVT'].values.astype(float)
    tw_gr = r2_tw_s_013['GR'].fillna(r2_tw_s_013['GR'].mean()).values.astype(float)
    r2_gr_all_078 = hw['GR'].interpolate(limit_direction='both').fillna(tw_gr.mean()).values.astype(float)
    hgr = r2_gr_all_078[r2_ev_slot_015.index]
    r2_beam_results_079 = [r2_beam_search_051(hgr, tw_tvt, tw_gr, last_tvt, bs, mc, es, r) for bs, mc, es, r in R2_BEAM_CONFIGS_005]
    r2_beam_mean_080 = np.stack(r2_beam_results_079, 0).mean(0)
    r2_out_050 = hw['TVT_input'].values.astype(float).copy()
    r2_out_050[list(r2_ev_slot_015.index)] = r2_beam_mean_080
    return r2_out_050

def r2_selector_well_code_081(hw):
    r2_eval_mask_082 = hw['TVT_input'].isna().to_numpy()
    r2_n_eval_083 = float(r2_eval_mask_082.sum())
    r2_z_eval_084 = hw.loc[r2_eval_mask_082, 'Z'].values.astype(float)
    r2_z_span_085 = float(np.nanmax(r2_z_eval_084) - np.nanmin(r2_z_eval_084)) if len(r2_z_eval_084) else 0.0
    r2_n_bin_086 = int(r2_n_eval_083 > SELECTOR_N_EVAL_THRESHOLD)
    r2_z_bin_087 = int(np.searchsorted(SELECTOR_Z_SPAN_THRESHOLDS, r2_z_span_085, side='right'))
    r2_code_088 = r2_n_bin_086 + 2 * r2_z_bin_087
    r2_variant_089 = SELECTOR_BIN_VARIANTS.get(r2_code_088, SELECTOR_GLOBAL_VARIANT)
    return (r2_code_088, r2_variant_089, r2_n_eval_083, r2_z_span_085)

def r2_parse_selector_variant_090(name):
    r2_parts_091 = name.split('_')
    scale = float(r2_parts_091[2])
    r2_beam_weight_092 = 0.0
    r2_hold_weight_093 = 0.0
    if 'beam' in r2_parts_091:
        r2_beam_weight_092 = float(r2_parts_091[r2_parts_091.index('beam') + 1])
    if 'hold' in r2_parts_091:
        r2_hold_weight_093 = float(r2_parts_091[r2_parts_091.index('hold') + 1])
    return (scale, r2_beam_weight_092, r2_hold_weight_093)

def apply_selector_variant(name, pf_by_scale, tvt_beam, last_known_tvt):
    scale, r2_beam_weight_092, r2_hold_weight_093 = r2_parse_selector_variant_090(name)
    r2_base_011 = pf_by_scale.get(f'pf_scale_{scale:g}')
    if r2_base_011 is None:
        r2_base_011 = pf_by_scale[SELECTOR_GLOBAL_VARIANT.split('_beam_')[0].split('_hold_')[0]]
    r2_pred_094 = (1.0 - r2_beam_weight_092) * r2_base_011 + r2_beam_weight_092 * tvt_beam
    r2_pred_094 = (1.0 - r2_hold_weight_093) * r2_pred_094 + r2_hold_weight_093 * last_known_tvt
    return r2_pred_094
```

```python
import numpy as _np, pandas as _pd

def _r2_robfit_401(s, y, deg=5):
    if len(s) < deg + 2:
        return y.copy()
    r2_c_slot_185 = _np.polyfit(s, y, deg)
    for _ in range(4):
        r = y - _np.polyval(r2_c_slot_185, s)
        r2_sc_slot_343 = _np.median(_np.abs(r)) * 1.4826 + 1e-06
        r2_c_slot_185 = _np.polyfit(s, y, deg, w=1.0 / (1.0 + (r / (2.0 * r2_sc_slot_343)) ** 2))
    return _np.polyval(r2_c_slot_185, s)
try:
    _r2_base_402 = _pd.read_csv('submission.csv')
    assert set(['id', 'tvt']).issubset(_r2_base_402.columns)
    _r2_base_402['well'] = _r2_base_402['id'].str[:8]
    _r2_base_402['row_idx'] = _r2_base_402['id'].str[9:].astype(int)
    _r2_out_403 = dict(zip(_r2_base_402['id'].values, _r2_base_402['tvt'].astype(float).values))
    _r2_n_ok_404 = 0
    for _r2_wid_405, _g in _r2_base_402.groupby('well'):
        try:
            _r2_hw_406 = _pd.read_csv(CFG.dataset_path / 'test' / (_r2_wid_405 + '__horizontal_well.csv'))
            _r2_kn_407 = _r2_hw_406[_r2_hw_406['TVT_input'].notna()]
            if len(_r2_kn_407) < 5:
                continue
            _r2_last_408 = _r2_kn_407.iloc[-1]
            _r2_anchor_409 = float(_r2_last_408['TVT_input']) + float(_r2_last_408['Z'])
            _r2_ps_410 = float(_r2_last_408['MD'])
            _r2_end_411 = float(_r2_hw_406['MD'].iloc[-1])
            _r2_gi_412 = _g.sort_values('row_idx')
            _r2_ri_413 = _r2_gi_412['row_idx'].values
            R2_Z_414 = _r2_hw_406['Z'].values[_r2_ri_413].astype(float)
            _r2_md_192 = _r2_hw_406['MD'].values[_r2_ri_413].astype(float)
            _r2_s_415 = (_r2_md_192 - _r2_ps_410) / max(_r2_end_411 - _r2_ps_410, 1e-06)
            _r2_tvt_416 = _r2_gi_412['tvt'].values.astype(float)
            _r2_fit_417 = _r2_robfit_401(_r2_s_415, _r2_tvt_416 + R2_Z_414 - _r2_anchor_409, 4)
            _r2_tvt_fit_full_418 = _r2_anchor_409 + _r2_fit_417 - R2_Z_414
            _r2_tvt_fit_419 = 0.25 * _r2_tvt_416 + 0.75 * _r2_tvt_fit_full_418
            if not _np.all(_np.isfinite(_r2_tvt_fit_419)):
                continue
            for _r2_rid_420, _r2_val_421 in zip(_r2_gi_412['id'].values, _r2_tvt_fit_419):
                _r2_out_403[_r2_rid_420] = float(_r2_val_421)
            _r2_n_ok_404 += 1
        except Exception as _r2_e_422:
            print('proj fallback', _r2_wid_405, _r2_e_422)
    print('projection applied to', _r2_n_ok_404, 'wells')
    _r2_final_423 = _r2_base_402[['id']].copy()
    _r2_final_423['tvt'] = _r2_final_423['id'].map(_r2_out_403).astype(float)
    _r2_final_423[['id', 'tvt']].to_csv('submission.csv', index=False)
    print('wrote projected submission.csv', _r2_final_423.shape)
except Exception as _r2_e_422:
    print('PROJECTION SKIPPED (kept blended submission):', _r2_e_422)
```

```python
class PP:
    alpha = 1.0
    tau = 85.0
    w_pf = 0.0
    w_sub1 = 0.6
    sub2_scale = 'scale_5'
    sg_win = 61
    sg_poly = 3

def r2_warmup_489(md_since, tau):
    return 1.0 - np.exp(-np.maximum(md_since, 0.0) / tau) if tau > 1e-06 else 1.0

def r2_make_prediction_490(df, model_delta, likpf):
    r2_last_016 = df['last_known_tvt'].values.astype(float)
    r2_pf_delta_491 = df['pf_ancc'].values.astype(float) - r2_last_016
    r2_lp_slot_159 = df[f'likpf_{PP.sub2_scale}'].values.astype(float) - r2_last_016
    r2_sub1_492 = PP.alpha * r2_warmup_489(df['md_since'].values.astype(float), PP.tau) * (model_delta * (1 - PP.w_pf) + r2_pf_delta_491 * PP.w_pf)
    r2_delta_493 = PP.w_sub1 * r2_sub1_492 + (1 - PP.w_sub1) * r2_lp_slot_159
    r2_pred_094 = r2_last_016 + r2_delta_493
    r2_out_050 = r2_pred_094.copy()
    r2_dfx_494 = df.reset_index(drop=True)
    for _, r2_idx_042 in r2_dfx_494.groupby('well', sort=False).groups.items():
        pos = r2_dfx_494.index.get_indexer(r2_idx_042)
        v = r2_pred_094[pos]
        r2_n_slot_052 = len(v)
        r2_wl_slot_372 = min(PP.sg_win, r2_n_slot_052)
        if r2_wl_slot_372 % 2 == 0:
            r2_wl_slot_372 -= 1
        if r2_wl_slot_372 >= PP.sg_poly + 2:
            r2_out_050[pos] = savgol_filter(v, r2_wl_slot_372, PP.sg_poly)
    return r2_out_050
```

```python
import os as _gold_os
import glob as _gold_glob
import json as _gold_json
import time as _gold_time
import hashlib as _gold_hashlib
from pathlib import Path as _GoldPath
import numpy as _gold_np
import pandas as _gold_pd
R2_GOLD_ENABLE_562 = _gold_os.environ.get('ROGII_GOLD_PREFIX_CAL', '1') == '1'
R2_GOLD_PROFILE_563 = _gold_os.environ.get('ROGII_GOLD_PROFILE', 'balanced').strip().lower()
R2_GOLD_INCLUDE_PF_564 = _gold_os.environ.get('ROGII_GOLD_INCLUDE_PF', '1') == '1'
R2_GOLD_CAL_SEEDS_565 = int(_gold_os.environ.get('ROGII_GOLD_CAL_SEEDS', '24'))
R2_GOLD_FINAL_SEEDS_566 = int(_gold_os.environ.get('ROGII_GOLD_FINAL_SEEDS', '48'))
R2_GOLD_PARTICLES_567 = int(_gold_os.environ.get('ROGII_GOLD_PARTICLES', '350'))
R2_GOLD_CUT_FRACS_568 = tuple((float(x) for x in _gold_os.environ.get('ROGII_GOLD_CUT_FRACS', '0.55,0.70,0.84').split(',') if x.strip()))
R2_GOLD_MAX_WELLS_569 = int(_gold_os.environ.get('ROGII_GOLD_MAX_WELLS', '1000000'))
R2_GOLD_PROFILES_570 = {'conservative': dict(min_gain=1.0, max_best=9.0, min_consistency=0.67, base=0.06, gain_scale=0.12, margin_scale=0.04, quality_bonus=0.02, cap=0.22, clip_base=8.0, clip_gain=3.0, clip_max=18.0, delta_soft=22.0, p95_hard=55.0), 'balanced': dict(min_gain=0.55, max_best=12.0, min_consistency=0.5, base=0.08, gain_scale=0.2, margin_scale=0.06, quality_bonus=0.04, cap=0.36, clip_base=10.0, clip_gain=4.5, clip_max=28.0, delta_soft=30.0, p95_hard=75.0), 'aggressive': dict(min_gain=0.25, max_best=15.0, min_consistency=0.34, base=0.12, gain_scale=0.32, margin_scale=0.1, quality_bonus=0.06, cap=0.56, clip_base=14.0, clip_gain=7.0, clip_max=45.0, delta_soft=42.0, p95_hard=110.0)}
if R2_GOLD_PROFILE_563 not in R2_GOLD_PROFILES_570:
    print(f'Unknown ROGII_GOLD_PROFILE={R2_GOLD_PROFILE_563!r}; using balanced')
    R2_GOLD_PROFILE_563 = 'balanced'

def _r2_cal_work_dir_571():
    return _GoldPath('/kaggle/working') if _GoldPath('/kaggle/working').exists() else _GoldPath('.')

def _r2_cal_find_data_572():
    r2_candidates_573 = []
    r2_obj_574 = globals().get('CFG')
    if r2_obj_574 is not None:
        for r2_attr_575 in ('dataset_path', 'DATA'):
            if hasattr(r2_obj_574, r2_attr_575):
                r2_candidates_573.append(_GoldPath(getattr(r2_obj_574, r2_attr_575)))
    r2_candidates_573.extend([_GoldPath('/kaggle/input/competitions/rogii-wellbore-geology-prediction'), _GoldPath('/kaggle/input/rogii-wellbore-geology-prediction'), _GoldPath('.')])
    for r2_c_slot_185 in r2_candidates_573:
        try:
            if (r2_c_slot_185 / 'train').exists() and (r2_c_slot_185 / 'test').exists() and (r2_c_slot_185 / 'sample_submission.csv').exists():
                return r2_c_slot_185
        except Exception:
            pass
    for p in _gold_glob.glob('/kaggle/input/**/sample_submission.csv', recursive=True):
        r2_c_slot_185 = _GoldPath(p).parent
        if (r2_c_slot_185 / 'train').exists() and (r2_c_slot_185 / 'test').exists():
            return r2_c_slot_185
    raise RuntimeError('Could not locate ROGII data directory')

def _r2_cal_split_ids_576(df):
    r2_out_050 = df.copy()
    r2_parts_091 = r2_out_050['id'].astype(str).str.rsplit('_', n=1, expand=True)
    if r2_parts_091.shape[1] != 2:
        raise RuntimeError('Unexpected id format; expected well_rowindex')
    r2_out_050['well'] = r2_parts_091[0]
    r2_out_050['row_idx'] = r2_parts_091[1].astype(int)
    return r2_out_050

def _r2_cal_rmse_577(a, b):
    a = _gold_np.asarray(a, dtype=float)
    b = _gold_np.asarray(b, dtype=float)
    r2_m_slot_023 = _gold_np.isfinite(a) & _gold_np.isfinite(b)
    if int(r2_m_slot_023.sum()) == 0:
        return float('inf')
    r2_d_slot_035 = a[r2_m_slot_023] - b[r2_m_slot_023]
    return float(_gold_np.sqrt(_gold_np.mean(r2_d_slot_035 * r2_d_slot_035)))

def _r2_cal_sha256_578(path):
    r2_h_slot_557 = _gold_hashlib.sha256()
    with open(path, 'rb') as r2_f_slot_382:
        for r2_chunk_558 in iter(lambda: r2_f_slot_382.read(1 << 20), b''):
            r2_h_slot_557.update(r2_chunk_558)
    return r2_h_slot_557.hexdigest()

def _r2_cal_robust_poly_predict_579(x_known, y_known, x_all, deg):
    x_known = _gold_np.asarray(x_known, dtype=float)
    y_known = _gold_np.asarray(y_known, dtype=float)
    x_all = _gold_np.asarray(x_all, dtype=float)
    r2_m_slot_023 = _gold_np.isfinite(x_known) & _gold_np.isfinite(y_known)
    x_known = x_known[r2_m_slot_023]
    y_known = y_known[r2_m_slot_023]
    if len(x_known) < 3:
        r2_fill_580 = float(_gold_np.nanmedian(y_known)) if len(y_known) else 0.0
        return _gold_np.full_like(x_all, r2_fill_580, dtype=float)
    deg = int(min(max(1, deg), len(x_known) - 1))
    r2_x0_slot_581 = float(x_known[0])
    r2_xs_slot_245 = float(_gold_np.nanmax(x_known) - _gold_np.nanmin(x_known))
    if not _gold_np.isfinite(r2_xs_slot_245) or r2_xs_slot_245 < 1e-06:
        r2_xs_slot_245 = 1.0
    r2_xk_slot_582 = (x_known - r2_x0_slot_581) / r2_xs_slot_245
    r2_xa_slot_583 = (x_all - r2_x0_slot_581) / r2_xs_slot_245
    try:
        r2_coef_241 = _gold_np.polyfit(r2_xk_slot_582, y_known, deg)
        for _ in range(5):
            r2_fit_584 = _gold_np.polyval(r2_coef_241, r2_xk_slot_582)
            r2_res_028 = y_known - r2_fit_584
            r2_sc_slot_343 = 1.4826 * float(_gold_np.nanmedian(_gold_np.abs(r2_res_028 - _gold_np.nanmedian(r2_res_028)))) + 1e-06
            r2_weights_048 = 1.0 / (1.0 + (r2_res_028 / (2.5 * r2_sc_slot_343)) ** 2)
            r2_coef_241 = _gold_np.polyfit(r2_xk_slot_582, y_known, deg, w=r2_weights_048)
        return _gold_np.polyval(r2_coef_241, r2_xa_slot_583).astype(float)
    except Exception:
        return _gold_np.full_like(x_all, float(_gold_np.nanmedian(y_known)), dtype=float)

def _r2_cal_variant_grid_585():
    variants = set()
    try:
        variants.update(SELECTOR_BIN_VARIANTS.values())
        variants.add(SELECTOR_GLOBAL_VARIANT)
    except Exception:
        pass
    for scale in (3, 5, 8, 12):
        for r2_hold_586 in (0.0, 0.05, 0.1, 0.15, 0.2, 0.25):
            variants.add(f'pf_scale_{scale:g}_hold_{r2_hold_586:g}')
        for r2_beam_587 in (0.05, 0.1, 0.2, 0.3):
            for r2_hold_586 in (0.0, 0.05, 0.1, 0.15, 0.2):
                variants.add(f'pf_scale_{scale:g}_beam_{r2_beam_587:g}_hold_{r2_hold_586:g}')
    return sorted(variants)

def _r2_cal_tvt_from_contacts_588(hw_tr, tw_tr, ref_col='EGFDU'):
    r2_tw_g_007 = tw_tr.dropna(subset=['Geology']) if 'Geology' in tw_tr.columns else tw_tr.copy()
    if 'Geology' in r2_tw_g_007.columns and ref_col in hw_tr.columns:
        r2_ref_tvt_008 = r2_tw_g_007.loc[r2_tw_g_007['Geology'] == ref_col, 'TVT'].min()
        if _gold_np.isnan(r2_ref_tvt_008):
            ref_col = r2_tw_g_007['Geology'].iloc[0]
            r2_ref_tvt_008 = r2_tw_g_007.loc[r2_tw_g_007['Geology'] == ref_col, 'TVT'].min()
        r2_offset_009 = (hw_tr['TVT'] - (r2_ref_tvt_008 - (hw_tr['Z'] - hw_tr[ref_col]))).mean()
        return (r2_ref_tvt_008 - (hw_tr['Z'] - hw_tr[ref_col]) + r2_offset_009).to_numpy(dtype=float)
    return hw_tr['TVT'].to_numpy(dtype=float)

def _r2_cal_contact_candidate_589(wid, hw, data_dir):
    r2_out_050 = {}
    try:
        r2_hw_tr_path_590 = data_dir / 'train' / f'{wid}__horizontal_well.csv'
        r2_tw_tr_path_591 = data_dir / 'train' / f'{wid}__typewell.csv'
        if not r2_hw_tr_path_590.exists() or not r2_tw_tr_path_591.exists():
            return r2_out_050
        hw_tr = _gold_pd.read_csv(r2_hw_tr_path_590)
        tw_tr = _gold_pd.read_csv(r2_tw_tr_path_591)
        r2_phys_592 = _r2_cal_tvt_from_contacts_588(hw_tr, tw_tr)
        md = hw_tr['MD'].to_numpy(dtype=float)
        r2_m_slot_023 = _gold_np.isfinite(md) & _gold_np.isfinite(r2_phys_592)
        if int(r2_m_slot_023.sum()) < 100:
            return r2_out_050
        r2_order_070 = _gold_np.argsort(md[r2_m_slot_023])
        r2_md_s_593 = md[r2_m_slot_023][r2_order_070]
        r2_ph_s_594 = r2_phys_592[r2_m_slot_023][r2_order_070]
        r2_pred_094 = _gold_np.interp(hw['MD'].to_numpy(dtype=float), r2_md_s_593, r2_ph_s_594, left=_gold_np.nan, right=_gold_np.nan)
        r2_out_050['contact_md_lookup'] = r2_pred_094.astype(float)
    except Exception:
        return r2_out_050
    return r2_out_050

def _r2_cal_poly_candidates_595(hw_masked):
    r2_out_050 = {}
    r2_tvt_560 = hw_masked['TVT_input'].to_numpy(dtype=float)
    md = hw_masked['MD'].to_numpy(dtype=float)
    r2_z_slot_596 = hw_masked['Z'].to_numpy(dtype=float)
    r2_kn_slot_014 = _gold_np.flatnonzero(_gold_np.isfinite(r2_tvt_560) & _gold_np.isfinite(md) & _gold_np.isfinite(r2_z_slot_596))
    if len(r2_kn_slot_014) < 30:
        return r2_out_050
    r2_u_slot_130 = r2_tvt_560 + r2_z_slot_596
    for r2_tail_019 in (80, 160, 320, 640, 1000000):
        r2_sel_597 = r2_kn_slot_014[-min(int(r2_tail_019), len(r2_kn_slot_014)):]
        if len(r2_sel_597) < 30:
            continue
        r2_tag_268 = 'all' if r2_tail_019 >= 1000000 else f'tail{r2_tail_019}'
        for deg in (1, 2, 3):
            if len(r2_sel_597) < deg + 12:
                continue
            r2_uhat_598 = _r2_cal_robust_poly_predict_579(md[r2_sel_597], r2_u_slot_130[r2_sel_597], md, deg)
            r2_out_050[f'poly_u_deg{deg}_{r2_tag_268}'] = (r2_uhat_598 - r2_z_slot_596).astype(float)
    return r2_out_050

def _r2_cal_surface_candidates_599(hw_masked, wid, data_dir):
    r2_out_050 = {}
    r2_tvt_560 = hw_masked['TVT_input'].to_numpy(dtype=float)
    r2_z_slot_596 = hw_masked['Z'].to_numpy(dtype=float)
    r2_xy_slot_227 = hw_masked[['X', 'Y']].to_numpy(dtype=float)
    r2_kn_slot_014 = _gold_np.isfinite(r2_tvt_560) & _gold_np.isfinite(r2_z_slot_596) & _gold_np.isfinite(r2_xy_slot_227).all(axis=1)
    if int(r2_kn_slot_014.sum()) < 30:
        return r2_out_050
    r2_formations_600 = list(globals().get('FORMATIONS', ['ANCC', 'ASTNU', 'ASTNL', 'EGFDU', 'EGFDL', 'BUDA']))
    r2_fi_slot_601 = globals().get('_FI', globals().get('FI', None))
    r2_di_slot_602 = globals().get('_DI', globals().get('DI', None))
    r2_surf_names_603 = []
    try:
        if r2_fi_slot_601 is not None:
            r2_form_all_604, _ = r2_fi_slot_601.impute(r2_xy_slot_227, self_wid=None)
            r2_form_all_604 = _gold_np.asarray(r2_form_all_604, dtype=float)
            for r2_i_slot_031, r2_fn_slot_237 in enumerate(r2_formations_600[:r2_form_all_604.shape[1]]):
                r2_f_slot_382 = r2_form_all_604[:, r2_i_slot_031]
                r2_good_605 = r2_kn_slot_014 & _gold_np.isfinite(r2_f_slot_382)
                if int(r2_good_605.sum()) < 30:
                    continue
                r2_b_med_606 = float(_gold_np.nanmedian(r2_tvt_560[r2_good_605] + r2_z_slot_596[r2_good_605] - r2_f_slot_382[r2_good_605]))
                r2_out_050[f'surface_{r2_fn_slot_237}_median'] = (-r2_z_slot_596 + r2_f_slot_382 + r2_b_med_606).astype(float)
                r2_surf_names_603.append(f'surface_{r2_fn_slot_237}_median')
                if callable(globals().get('seg_b_well')):
                    try:
                        r2_b_full_199, _, _, r2_b_late_200, r2_b_wls_204 = seg_b_well(r2_tvt_560[r2_good_605].astype(_gold_np.float32), r2_z_slot_596[r2_good_605].astype(_gold_np.float32), r2_f_slot_382[r2_good_605].astype(_gold_np.float32))
                        r2_out_050[f'surface_{r2_fn_slot_237}_full'] = (-r2_z_slot_596 + r2_f_slot_382 + float(r2_b_full_199)).astype(float)
                        r2_out_050[f'surface_{r2_fn_slot_237}_late'] = (-r2_z_slot_596 + r2_f_slot_382 + float(r2_b_late_200)).astype(float)
                        r2_out_050[f'surface_{r2_fn_slot_237}_wls'] = (-r2_z_slot_596 + r2_f_slot_382 + float(r2_b_wls_204)).astype(float)
                        r2_surf_names_603.extend([f'surface_{r2_fn_slot_237}_full', f'surface_{r2_fn_slot_237}_late', f'surface_{r2_fn_slot_237}_wls'])
                    except Exception:
                        pass
    except Exception as r2_e_slot_388:
        print('surface imputer skipped', wid, r2_e_slot_388)
    try:
        if r2_di_slot_602 is not None:
            r2_dense_607, _, _ = r2_di_slot_602.impute(r2_xy_slot_227, self_wid=None)
            r2_dense_607 = _gold_np.asarray(r2_dense_607, dtype=float)
            r2_good_605 = r2_kn_slot_014 & _gold_np.isfinite(r2_dense_607)
            if int(r2_good_605.sum()) >= 30:
                r2_b_med_606 = float(_gold_np.nanmedian(r2_tvt_560[r2_good_605] + r2_z_slot_596[r2_good_605] - r2_dense_607[r2_good_605]))
                r2_out_050['dense_ancc_median'] = (-r2_z_slot_596 + r2_dense_607 + r2_b_med_606).astype(float)
                r2_surf_names_603.append('dense_ancc_median')
                if callable(globals().get('seg_b_well')):
                    try:
                        r2_b_full_199, _, _, r2_b_late_200, r2_b_wls_204 = seg_b_well(r2_tvt_560[r2_good_605].astype(_gold_np.float32), r2_z_slot_596[r2_good_605].astype(_gold_np.float32), r2_dense_607[r2_good_605].astype(_gold_np.float32))
                        r2_out_050['dense_ancc_full'] = (-r2_z_slot_596 + r2_dense_607 + float(r2_b_full_199)).astype(float)
                        r2_out_050['dense_ancc_late'] = (-r2_z_slot_596 + r2_dense_607 + float(r2_b_late_200)).astype(float)
                        r2_out_050['dense_ancc_wls'] = (-r2_z_slot_596 + r2_dense_607 + float(r2_b_wls_204)).astype(float)
                        r2_surf_names_603.extend(['dense_ancc_full', 'dense_ancc_late', 'dense_ancc_wls'])
                    except Exception:
                        pass
    except Exception as r2_e_slot_388:
        print('dense imputer skipped', wid, r2_e_slot_388)
    r2_ens_names_608 = [r2_n_slot_052 for r2_n_slot_052 in r2_surf_names_603 if r2_n_slot_052 in r2_out_050]
    if len(r2_ens_names_608) >= 2:
        r2_errs_609 = _gold_np.array([_r2_cal_rmse_577(r2_out_050[r2_n_slot_052][r2_kn_slot_014], r2_tvt_560[r2_kn_slot_014]) for r2_n_slot_052 in r2_ens_names_608], dtype=float)
        r2_finite_610 = _gold_np.isfinite(r2_errs_609)
        if int(r2_finite_610.sum()) >= 2:
            r2_names_507 = [r2_n_slot_052 for r2_n_slot_052, r2_ok_slot_611 in zip(r2_ens_names_608, r2_finite_610) if r2_ok_slot_611]
            r2_errs_609 = r2_errs_609[r2_finite_610]
            r2_weights_048 = 1.0 / _gold_np.maximum(r2_errs_609, 0.25) ** 2
            r2_weights_048 = r2_weights_048 / r2_weights_048.sum()
            r2_mat_612 = _gold_np.vstack([r2_out_050[r2_n_slot_052] for r2_n_slot_052 in r2_names_507])
            r2_out_050['surface_weighted_prefix'] = (r2_weights_048[:, None] * r2_mat_612).sum(axis=0).astype(float)
    r2_out_050.update(_r2_cal_contact_candidate_589(wid, hw_masked, data_dir))
    return r2_out_050

def _r2_cal_pf_candidates_613(hw_masked, tw, variants, n_seeds, n_particles):
    r2_out_050 = {}
    if not R2_GOLD_INCLUDE_PF_564:
        return r2_out_050
    if not callable(globals().get('run_pf_lik_ensemble_scales')) or not callable(globals().get('apply_selector_variant')):
        return r2_out_050
    r2_kn_slot_014 = hw_masked[hw_masked['TVT_input'].notna()]
    r2_ev_slot_015 = hw_masked[hw_masked['TVT_input'].isna()]
    if len(r2_kn_slot_014) < 30 or len(r2_ev_slot_015) == 0:
        return r2_out_050
    try:
        pf_by_scale = run_pf_lik_ensemble_scales(hw_masked, tw, scales=tuple(globals().get('SELECTOR_SCALES', (3.0, 5.0, 8.0, 12.0))), n_particles=int(n_particles), n_seeds=int(n_seeds))
        try:
            tvt_beam = run_beam_ensemble(hw_masked, tw)
        except Exception:
            tvt_beam = pf_by_scale.get('pf_mean')
            if tvt_beam is None:
                tvt_beam = next(iter(pf_by_scale.values()))
        last_known_tvt = float(r2_kn_slot_014['TVT_input'].iloc[-1])
        for r2_variant_089 in variants:
            try:
                r2_pred_094 = apply_selector_variant(r2_variant_089, pf_by_scale, tvt_beam, last_known_tvt)
                if r2_pred_094 is not None and len(r2_pred_094) == len(hw_masked):
                    r2_out_050['pf|' + r2_variant_089] = _gold_np.asarray(r2_pred_094, dtype=float)
            except Exception:
                pass
    except Exception as r2_e_slot_388:
        print('PF calibration skipped:', r2_e_slot_388)
    return r2_out_050

def _r2_cal_candidate_pool_614(wid, hw_masked, tw, data_dir, variants, include_pf=True, n_seeds=24, n_particles=350):
    r2_pool_615 = {}
    r2_pool_615.update(_r2_cal_poly_candidates_595(hw_masked))
    r2_pool_615.update(_r2_cal_surface_candidates_599(hw_masked, wid, data_dir))
    if include_pf:
        r2_pool_615.update(_r2_cal_pf_candidates_613(hw_masked, tw, variants, n_seeds=n_seeds, n_particles=n_particles))
    r2_clean_616 = {}
    for name, r2_pred_094 in r2_pool_615.items():
        arr = _gold_np.asarray(r2_pred_094, dtype=float)
        if len(arr) == len(hw_masked) and _gold_np.isfinite(arr).sum() >= max(20, len(hw_masked) // 20):
            r2_clean_616[name] = arr
    return r2_clean_616

def _r2_cal_default_pf_name_617(hw):
    try:
        return 'pf|' + r2_selector_well_code_081(hw)[1]
    except Exception:
        try:
            return 'pf|' + SELECTOR_GLOBAL_VARIANT
        except Exception:
            return None

def _r2_cal_calibrate_well_618(wid, hw, tw, data_dir, variants):
    r2_tvt_560 = hw['TVT_input'].to_numpy(dtype=float)
    r2_is_known_619 = _gold_np.isfinite(r2_tvt_560)
    r2_is_hidden_620 = ~r2_is_known_619
    if not bool(r2_is_hidden_620.any()):
        return None
    r2_first_hidden_621 = int(_gold_np.flatnonzero(r2_is_hidden_620)[0])
    r2_known_prefix_622 = _gold_np.flatnonzero(r2_is_known_619 & (_gold_np.arange(len(hw)) < r2_first_hidden_621))
    if len(r2_known_prefix_622) < 140:
        return dict(well=wid, status='skip_short_prefix', known_prefix=int(len(r2_known_prefix_622)))
    r2_cuts_623 = []
    for r2_frac_342 in R2_GOLD_CUT_FRACS_568:
        r2_cut_pos_624 = int(round(len(r2_known_prefix_622) * float(r2_frac_342)))
        r2_cut_pos_624 = max(50, min(r2_cut_pos_624, len(r2_known_prefix_622) - 35))
        if r2_cut_pos_624 <= 0 or r2_cut_pos_624 >= len(r2_known_prefix_622):
            continue
        r2_cutoff_idx_625 = int(r2_known_prefix_622[r2_cut_pos_624 - 1])
        r2_hold_idx_626 = r2_known_prefix_622[r2_cut_pos_624:]
        if len(r2_hold_idx_626) >= 35:
            r2_cuts_623.append((float(r2_frac_342), r2_cutoff_idx_625, r2_hold_idx_626))
    if not r2_cuts_623:
        return dict(well=wid, status='skip_no_holdout', known_prefix=int(len(r2_known_prefix_622)))
    r2_scores_221 = {}
    r2_cut_rows_627 = []
    r2_default_name_628 = _r2_cal_default_pf_name_617(hw)
    for r2_frac_342, r2_cutoff_idx_625, r2_hold_idx_626 in r2_cuts_623:
        r2_hw_m_629 = hw.copy(deep=True)
        r2_hw_m_629.loc[r2_hw_m_629.index > r2_cutoff_idx_625, 'TVT_input'] = _gold_np.nan
        r2_pool_615 = _r2_cal_candidate_pool_614(wid, r2_hw_m_629, tw, data_dir, variants, include_pf=R2_GOLD_INCLUDE_PF_564, n_seeds=R2_GOLD_CAL_SEEDS_565, n_particles=R2_GOLD_PARTICLES_567)
        y = r2_tvt_560[r2_hold_idx_626]
        r2_row_226 = {'well': wid, 'cut_frac': r2_frac_342, 'holdout_rows': int(len(r2_hold_idx_626)), 'candidates': int(len(r2_pool_615))}
        r2_local_630 = []
        for name, r2_pred_094 in r2_pool_615.items():
            r2_err_631 = _r2_cal_rmse_577(r2_pred_094[r2_hold_idx_626], y)
            if _gold_np.isfinite(r2_err_631):
                r2_scores_221.setdefault(name, []).append(r2_err_631)
                r2_local_630.append((r2_err_631, name))
        r2_local_630.sort()
        if r2_local_630:
            r2_row_226['best_name'] = r2_local_630[0][1]
            r2_row_226['best_rmse'] = float(r2_local_630[0][0])
            if r2_default_name_628 in r2_pool_615:
                r2_row_226['default_rmse'] = float(_r2_cal_rmse_577(r2_pool_615[r2_default_name_628][r2_hold_idx_626], y))
            else:
                r2_row_226['default_rmse'] = float('nan')
        r2_cut_rows_627.append(r2_row_226)
    if not r2_scores_221:
        return dict(well=wid, status='skip_no_scores', known_prefix=int(len(r2_known_prefix_622)))
    r2_agg_632 = {}
    for name, vals in r2_scores_221.items():
        vals = _gold_np.asarray(vals, dtype=float)
        r2_agg_632[name] = float(_gold_np.nanmedian(vals) + 0.1 * _gold_np.nanstd(vals))
    r2_ordered_633 = sorted(((v, k) for k, v in r2_agg_632.items() if _gold_np.isfinite(v)))
    if not r2_ordered_633:
        return dict(well=wid, status='skip_nonfinite_scores', known_prefix=int(len(r2_known_prefix_622)))
    r2_best_score_634, r2_best_name_635 = r2_ordered_633[0]
    r2_second_score_636 = r2_ordered_633[1][0] if len(r2_ordered_633) > 1 else r2_best_score_634
    if r2_default_name_628 is not None and r2_default_name_628 in r2_agg_632:
        r2_default_score_637 = float(r2_agg_632[r2_default_name_628])
    else:
        r2_pf_scores_638 = [v for k, v in r2_agg_632.items() if k.startswith('pf|')]
        r2_default_score_637 = float(_gold_np.nanmedian(r2_pf_scores_638)) if r2_pf_scores_638 else float(r2_second_score_636)
    r2_consistency_639 = 0.0
    r2_comparable_640 = 0
    for r2_row_226 in r2_cut_rows_627:
        if _gold_np.isfinite(r2_row_226.get('default_rmse', _gold_np.nan)):
            r2_comparable_640 += 1
            if r2_row_226.get('best_rmse', float('inf')) <= r2_row_226['default_rmse'] - 0.25:
                r2_consistency_639 += 1.0
    if r2_comparable_640:
        r2_consistency_639 /= r2_comparable_640
    else:
        r2_winners_641 = [r.get('best_name') for r in r2_cut_rows_627 if r.get('best_name')]
        r2_consistency_639 = float(sum((w == r2_best_name_635 for w in r2_winners_641))) / max(1, len(r2_winners_641))
    return dict(well=wid, status='ok', known_prefix=int(len(r2_known_prefix_622)), cuts=int(len(r2_cut_rows_627)), candidate_count=int(len(r2_agg_632)), best_name=r2_best_name_635, best_score=float(r2_best_score_634), second_score=float(r2_second_score_636), default_name=r2_default_name_628, default_score=float(r2_default_score_637), gain=float(r2_default_score_637 - r2_best_score_634), rank_margin=float(r2_second_score_636 - r2_best_score_634), consistency=float(r2_consistency_639), cut_rows=r2_cut_rows_627)

def _r2_cal_alpha_642(report, delta_rmse, delta_p95, profile_name):
    p = R2_GOLD_PROFILES_570[profile_name]
    if report.get('status') != 'ok':
        return 0.0
    r2_gain_643 = float(report.get('gain', 0.0))
    r2_best_143 = float(report.get('best_score', float('inf')))
    r2_margin_644 = float(report.get('rank_margin', 0.0))
    r2_consistency_639 = float(report.get('consistency', 0.0))
    if not _gold_np.isfinite(r2_best_143) or r2_best_143 > p['max_best'] or r2_gain_643 < p['min_gain'] or (r2_consistency_639 < p['min_consistency']):
        return 0.0
    alpha = p['base']
    alpha += p['gain_scale'] * min(max(r2_gain_643, 0.0), 5.0) / 5.0
    alpha += p['margin_scale'] * min(max(r2_margin_644, 0.0), 3.0) / 3.0
    if r2_best_143 <= 5.0:
        alpha += p['quality_bonus']
    r2_best_name_635 = str(report.get('best_name', ''))
    if (r2_best_name_635.startswith('surface_') or r2_best_name_635.startswith('dense_') or r2_best_name_635.startswith('poly_') or r2_best_name_635.startswith('contact_')) and r2_consistency_639 >= 0.67:
        alpha += 0.03 if profile_name != 'aggressive' else 0.06
    if _gold_np.isfinite(delta_rmse) and delta_rmse > p['delta_soft']:
        alpha *= max(0.2, p['delta_soft'] / max(delta_rmse, 1e-06))
    if _gold_np.isfinite(delta_p95) and delta_p95 > p['p95_hard']:
        return 0.0
    return float(min(p['cap'], max(0.0, alpha)))

def _r2_cal_profile_output_645(base_sub, candidate_by_id, reports_by_well, profile_name):
    r2_prof_646 = R2_GOLD_PROFILES_570[profile_name]
    r2_out_050 = base_sub.copy()
    r2_move_rows_647 = []
    for wid, r2_rep_648 in reports_by_well.items():
        r2_ids_649 = r2_out_050.loc[r2_out_050['well'] == wid, 'id'].astype(str).tolist()
        if not r2_ids_649:
            continue
        r2_cand_002 = _gold_np.array([candidate_by_id.get(r2_i_slot_031, _gold_np.nan) for r2_i_slot_031 in r2_ids_649], dtype=float)
        r2_idx_042 = r2_out_050.index[r2_out_050['well'] == wid].to_numpy()
        r2_base_011 = r2_out_050.loc[r2_idx_042, 'tvt'].to_numpy(dtype=float)
        r2_ok_slot_611 = _gold_np.isfinite(r2_cand_002) & _gold_np.isfinite(r2_base_011)
        if int(r2_ok_slot_611.sum()) != len(r2_base_011):
            r2_rep_648 = dict(r2_rep_648)
            r2_rep_648['apply_status'] = 'skip_nonfinite_candidate'
            r2_move_rows_647.append(r2_rep_648)
            continue
        r2_diff_520 = r2_cand_002 - r2_base_011
        delta_rmse = float(_gold_np.sqrt(_gold_np.mean(r2_diff_520 * r2_diff_520))) if len(r2_diff_520) else float('nan')
        delta_p95 = float(_gold_np.quantile(_gold_np.abs(r2_diff_520), 0.95)) if len(r2_diff_520) else float('nan')
        alpha = _r2_cal_alpha_642(r2_rep_648, delta_rmse, delta_p95, profile_name)
        r2_gain_643 = max(0.0, float(r2_rep_648.get('gain', 0.0)))
        r2_max_move_650 = min(r2_prof_646['clip_max'], r2_prof_646['clip_base'] + r2_prof_646['clip_gain'] * _gold_np.sqrt(r2_gain_643 + 1e-09))
        r2_ramp_651 = 1.0 - _gold_np.exp(-_gold_np.arange(len(r2_diff_520), dtype=float) / max(80.0, 0.12 * max(1, len(r2_diff_520))))
        r2_move_652 = _gold_np.clip(alpha * r2_ramp_651 * r2_diff_520, -r2_max_move_650, r2_max_move_650)
        r2_out_050.loc[r2_idx_042, 'tvt'] = r2_base_011 + r2_move_652
        r2_row_226 = dict(r2_rep_648)
        r2_row_226.update(dict(profile=profile_name, alpha=float(alpha), delta_rmse_vs_public=float(delta_rmse), delta_p95_vs_public=float(delta_p95), max_move_clip=float(r2_max_move_650), applied_rows=int(len(r2_idx_042)), mean_abs_move=float(_gold_np.mean(_gold_np.abs(r2_move_652))) if len(r2_move_652) else 0.0, max_abs_move=float(_gold_np.max(_gold_np.abs(r2_move_652))) if len(r2_move_652) else 0.0, apply_status='applied' if alpha > 0 else 'kept_public'))
        r2_move_rows_647.append(r2_row_226)
    return (r2_out_050, r2_move_rows_647)

def _r2_cal_reapply_guarded_contact_override_653(sub_df, data_dir):
    sub = _r2_cal_split_ids_576(sub_df[['id', 'tvt']])
    r2_pred_094 = dict(zip(sub['id'].astype(str), sub['tvt'].astype(float)))
    r2_train_wells_381 = set((p.stem.replace('__horizontal_well', '') for p in (data_dir / 'train').glob('*__horizontal_well.csv')))
    r2_n_ok_654 = 0
    r2_n_skip_655 = 0
    for wid, r2_g_slot_353 in sub.groupby('well'):
        if wid not in r2_train_wells_381:
            continue
        try:
            r2_hw_te_385 = _gold_pd.read_csv(data_dir / 'test' / f'{wid}__horizontal_well.csv')
            hw_tr = _gold_pd.read_csv(data_dir / 'train' / f'{wid}__horizontal_well.csv')
            tw_tr = _gold_pd.read_csv(data_dir / 'train' / f'{wid}__typewell.csv')
            r2_phys_592 = _r2_cal_tvt_from_contacts_588(hw_tr, tw_tr)
            r2_md_raw_656 = hw_tr['MD'].to_numpy(dtype=float)
            r2_m_slot_023 = _gold_np.isfinite(r2_phys_592) & _gold_np.isfinite(r2_md_raw_656)
            if int(r2_m_slot_023.sum()) < 100:
                r2_n_skip_655 += 1
                continue
            r2_order_070 = _gold_np.argsort(r2_md_raw_656[r2_m_slot_023])
            r2_md_tr_657 = r2_md_raw_656[r2_m_slot_023][r2_order_070]
            r2_ph_tr_658 = r2_phys_592[r2_m_slot_023][r2_order_070]
            r2_kn_slot_014 = r2_hw_te_385[r2_hw_te_385['TVT_input'].notna()]
            r2_kn_slot_014 = r2_kn_slot_014[(r2_kn_slot_014['MD'] >= r2_md_tr_657[0]) & (r2_kn_slot_014['MD'] <= r2_md_tr_657[-1])]
            if len(r2_kn_slot_014) < 50:
                r2_n_skip_655 += 1
                continue
            r2_rk_slot_659 = _r2_cal_rmse_577(_gold_np.interp(r2_kn_slot_014['MD'].to_numpy(dtype=float), r2_md_tr_657, r2_ph_tr_658), r2_kn_slot_014['TVT_input'].to_numpy(dtype=float))
            if not _gold_np.isfinite(r2_rk_slot_659) or r2_rk_slot_659 > 1.0:
                r2_n_skip_655 += 1
                continue
            r2_md_te_660 = r2_hw_te_385['MD'].to_numpy(dtype=float)
            r2_n_row_661 = 0
            for r2_rid_486, r2_ri_slot_662 in zip(r2_g_slot_353['id'].astype(str).values, r2_g_slot_353['row_idx'].values):
                r2_ri_slot_662 = int(r2_ri_slot_662)
                if 0 <= r2_ri_slot_662 < len(r2_md_te_660):
                    r2_mte_663 = float(r2_md_te_660[r2_ri_slot_662])
                    if r2_md_tr_657[0] <= r2_mte_663 <= r2_md_tr_657[-1]:
                        r2_pred_094[r2_rid_486] = float(_gold_np.interp(r2_mte_663, r2_md_tr_657, r2_ph_tr_658))
                        r2_n_row_661 += 1
            print('gold contact override OK %s rmse=%.4f rows=%d/%d' % (wid, r2_rk_slot_659, r2_n_row_661, len(r2_g_slot_353)))
            r2_n_ok_654 += 1
        except Exception as r2_e_slot_388:
            print('gold contact override fallback', wid, r2_e_slot_388)
            r2_n_skip_655 += 1
    sub['tvt'] = sub['id'].astype(str).map(r2_pred_094).astype(float)
    print('gold contact override done: overridden=%d skipped=%d' % (r2_n_ok_654, r2_n_skip_655))
    return sub[['id', 'tvt']]

def _r2_cal_validate_and_write_664(sub, sample, path):
    r2_out_050 = sub[['id', 'tvt']].copy()
    r2_out_050['id'] = r2_out_050['id'].astype(str)
    r2_out_050['tvt'] = r2_out_050['tvt'].astype(float)
    if list(r2_out_050.columns) != ['id', 'tvt']:
        raise RuntimeError('bad output columns')
    if len(r2_out_050) != len(sample):
        raise RuntimeError('bad output length')
    if not r2_out_050['id'].equals(sample['id'].astype(str)):
        raise RuntimeError('id order mismatch')
    if not _gold_np.isfinite(r2_out_050['tvt'].to_numpy(dtype=float)).all():
        raise RuntimeError('non-finite tvt in output')
    r2_out_050.to_csv(path, index=False)
    return r2_out_050
if not R2_GOLD_ENABLE_562:
    print('Gold visible-prefix calibration disabled; keeping public submission.csv')
else:
    _r2_cal_t0_665 = _gold_time.time()
    R2_GOLD_WORK_666 = _r2_cal_work_dir_571()
    R2_GOLD_DATA_667 = _r2_cal_find_data_572()
    _r2_cal_sample_668 = _gold_pd.read_csv(R2_GOLD_DATA_667 / 'sample_submission.csv')[['id']].copy()
    _r2_cal_sample_668['id'] = _r2_cal_sample_668['id'].astype(str)
    _r2_cal_base_669 = _gold_pd.read_csv(R2_GOLD_WORK_666 / 'submission.csv')[['id', 'tvt']].copy()
    _r2_cal_base_669['id'] = _r2_cal_base_669['id'].astype(str)
    _r2_cal_base_669['tvt'] = _r2_cal_base_669['tvt'].astype(float)
    _r2_cal_validate_and_write_664(_r2_cal_base_669, _r2_cal_sample_668, R2_GOLD_WORK_666 / 'submission_public_self_verified.csv')
    _r2_cal_base_669 = _r2_cal_split_ids_576(_r2_cal_base_669)
    _r2_cal_variants_670 = _r2_cal_variant_grid_585()
    print('Gold visible-prefix calibration:', dict(profile=R2_GOLD_PROFILE_563, include_pf=R2_GOLD_INCLUDE_PF_564, cal_seeds=R2_GOLD_CAL_SEEDS_565, final_seeds=R2_GOLD_FINAL_SEEDS_566, particles=R2_GOLD_PARTICLES_567, cut_fracs=R2_GOLD_CUT_FRACS_568, variants=len(_r2_cal_variants_670)))
    _r2_cal_reports_671 = []
    _r2_cal_cut_reports_672 = []
    _r2_cal_candidate_by_id_673 = {}
    _r2_cal_wells_674 = list(_r2_cal_base_669['well'].drop_duplicates())[:R2_GOLD_MAX_WELLS_569]
    for _r2_wi_675, _r2_wid_405 in enumerate(_r2_cal_wells_674, 1):
        try:
            _r2_hw_path_676 = R2_GOLD_DATA_667 / 'test' / f'{_r2_wid_405}__horizontal_well.csv'
            _r2_tw_path_677 = R2_GOLD_DATA_667 / 'test' / f'{_r2_wid_405}__typewell.csv'
            if not _r2_hw_path_676.exists() or not _r2_tw_path_677.exists():
                _r2_cal_reports_671.append(dict(well=_r2_wid_405, status='skip_missing_files'))
                continue
            _r2_hw_406 = _gold_pd.read_csv(_r2_hw_path_676)
            _r2_tw_678 = _gold_pd.read_csv(_r2_tw_path_677)
            print('[gold %d/%d] calibrating %s' % (_r2_wi_675, len(_r2_cal_wells_674), _r2_wid_405), flush=True)
            _r2_rep_679 = _r2_cal_calibrate_well_618(_r2_wid_405, _r2_hw_406, _r2_tw_678, R2_GOLD_DATA_667, _r2_cal_variants_670)
            if _r2_rep_679 is None:
                _r2_rep_679 = dict(well=_r2_wid_405, status='skip_none')
            _r2_cut_rows_680 = _r2_rep_679.pop('cut_rows', []) if isinstance(_r2_rep_679, dict) else []
            _r2_cal_cut_reports_672.extend(_r2_cut_rows_680)
            if _r2_rep_679.get('status') == 'ok':
                _r2_best_name_681 = _r2_rep_679['best_name']
                _r2_need_pf_final_682 = str(_r2_best_name_681).startswith('pf|')
                _r2_pool_final_683 = _r2_cal_candidate_pool_614(_r2_wid_405, _r2_hw_406, _r2_tw_678, R2_GOLD_DATA_667, _r2_cal_variants_670, include_pf=_r2_need_pf_final_682, n_seeds=R2_GOLD_FINAL_SEEDS_566, n_particles=R2_GOLD_PARTICLES_567)
                if _r2_best_name_681 not in _r2_pool_final_683 and _r2_need_pf_final_682:
                    _r2_pool_final_683 = _r2_cal_candidate_pool_614(_r2_wid_405, _r2_hw_406, _r2_tw_678, R2_GOLD_DATA_667, _r2_cal_variants_670, include_pf=False, n_seeds=0, n_particles=R2_GOLD_PARTICLES_567)
                if _r2_best_name_681 in _r2_pool_final_683:
                    _g = _r2_cal_base_669[_r2_cal_base_669['well'] == _r2_wid_405]
                    _r2_arr_684 = _r2_pool_final_683[_r2_best_name_681]
                    for _r2_rid_420, _r2_ri_413 in zip(_g['id'].astype(str).values, _g['row_idx'].astype(int).values):
                        if 0 <= int(_r2_ri_413) < len(_r2_arr_684) and _gold_np.isfinite(_r2_arr_684[int(_r2_ri_413)]):
                            _r2_cal_candidate_by_id_673[_r2_rid_420] = float(_r2_arr_684[int(_r2_ri_413)])
                    _r2_rep_679['final_candidate_available'] = True
                else:
                    _r2_rep_679['final_candidate_available'] = False
                    _r2_rep_679['status'] = 'skip_no_final_candidate'
            _r2_cal_reports_671.append(_r2_rep_679)
            print('  report:', {k: _r2_rep_679.get(k) for k in ['status', 'best_name', 'best_score', 'default_score', 'gain', 'consistency']}, flush=True)
        except Exception as _r2_e_422:
            print('gold calibration fallback', _r2_wid_405, _r2_e_422)
            _r2_cal_reports_671.append(dict(well=_r2_wid_405, status='error', error=str(_r2_e_422)))
    _r2_cal_report_df_685 = _gold_pd.DataFrame(_r2_cal_reports_671)
    _r2_cal_report_df_685.to_csv(R2_GOLD_WORK_666 / 'gold_prefix_calibration_report.csv', index=False)
    if _r2_cal_cut_reports_672:
        _gold_pd.DataFrame(_r2_cal_cut_reports_672).to_csv(R2_GOLD_WORK_666 / 'gold_prefix_cut_report.csv', index=False)
    _r2_reports_by_well_686 = {r.get('well'): r for r in _r2_cal_reports_671 if isinstance(r, dict) and r.get('well')}
    _r2_profile_summaries_687 = {}
    for _r2_profile_name_688 in ('conservative', 'balanced', 'aggressive'):
        _r2_profile_sub_689, _r2_move_rows_690 = _r2_cal_profile_output_645(_r2_cal_base_669, _r2_cal_candidate_by_id_673, _r2_reports_by_well_686, _r2_profile_name_688)
        _r2_profile_sub_689 = _r2_cal_reapply_guarded_contact_override_653(_r2_profile_sub_689[['id', 'tvt']], R2_GOLD_DATA_667)
        _r2_path_691 = R2_GOLD_WORK_666 / f'submission_gold_prefix_{_r2_profile_name_688}.csv'
        _r2_profile_sub_689 = _r2_cal_validate_and_write_664(_r2_profile_sub_689, _r2_cal_sample_668, _r2_path_691)
        _r2_move_df_692 = _gold_pd.DataFrame(_r2_move_rows_690)
        _r2_move_df_692.to_csv(R2_GOLD_WORK_666 / f'gold_prefix_moves_{_r2_profile_name_688}.csv', index=False)
        _r2_profile_summaries_687[_r2_profile_name_688] = dict(file=str(_r2_path_691), rows=int(len(_r2_profile_sub_689)), sha256=_r2_cal_sha256_578(_r2_path_691), applied_wells=int((_r2_move_df_692.get('apply_status') == 'applied').sum()) if 'apply_status' in _r2_move_df_692 else 0, mean_abs_move=float(_r2_move_df_692['mean_abs_move'].mean()) if 'mean_abs_move' in _r2_move_df_692 and len(_r2_move_df_692) else 0.0, max_abs_move=float(_r2_move_df_692['max_abs_move'].max()) if 'max_abs_move' in _r2_move_df_692 and len(_r2_move_df_692) else 0.0)
    _r2_chosen_path_693 = R2_GOLD_WORK_666 / f'submission_gold_prefix_{R2_GOLD_PROFILE_563}.csv'
    _r2_chosen_694 = _gold_pd.read_csv(_r2_chosen_path_693)
    _r2_chosen_694 = _r2_cal_validate_and_write_664(_r2_chosen_694, _r2_cal_sample_668, R2_GOLD_WORK_666 / 'submission.csv')
    _r2_audit_561 = dict(selected_profile=R2_GOLD_PROFILE_563, selected_sha256=_r2_cal_sha256_578(R2_GOLD_WORK_666 / 'submission.csv'), public_anchor_sha256=_r2_cal_sha256_578(R2_GOLD_WORK_666 / 'submission_public_self_verified.csv'), elapsed_sec=float(_gold_time.time() - _r2_cal_t0_665), wells=int(len(_r2_cal_wells_674)), candidates_with_final_values=int(len(_r2_cal_candidate_by_id_673)), profiles=_r2_profile_summaries_687)
    with open(R2_GOLD_WORK_666 / 'gold_prefix_submission_audit.json', 'w', encoding='utf-8') as r2_f_slot_382:
        _gold_json.dump(_r2_audit_561, r2_f_slot_382, indent=2, sort_keys=True)
    print('Gold visible-prefix selected submission.csv:', _r2_audit_561, flush=True)
```
